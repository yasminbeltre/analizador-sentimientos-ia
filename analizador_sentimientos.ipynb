
# 💬 Analizador de Sentimientos con IA

Proyecto de análisis de sentimientos en español utilizando Hugging Face Transformers.
Aplicación orientada a Customer Success, experiencia del cliente y análisis de feedback.

## 🤖 Carga del modelo

```python
from transformers import pipeline

analizador = pipeline("sentiment-analysis")
```

## 🔍 Análisis de texto individual

```python
texto = "Me encantó este curso de inteligencia artificial"
resultado = analizador(texto)
print(resultado)
```

## ✍️ Entrada de texto por el usuario

```python
texto = input("Escribe un texto: ")
resultado = analizador(texto)
print(resultado)
```

## 📊 Análisis de múltiples comentarios

```python
comentarios = [
    "Me encantó el servicio y la atención",
    "La experiencia fue terrible",
    "El producto llegó a tiempo",
    "No volvería a comprar",
    "Todo salió según lo esperado"
]

for comentario in comentarios:
    print(comentario, "->", analizador(comentario)[0])
```

## 📈 Visualización de resultados

```python
import matplotlib.pyplot as plt

categorias = ["Positivos", "Negativos", "Neutros"]
valores = [2, 2, 1]

plt.figure(figsize=(6,4))
plt.bar(categorias, valores)
plt.title("Distribución de Sentimientos")
plt.show()
```

## 🗂️ Exportación de resultados

```python
import pandas as pd

datos = {
    "Comentarios": comentarios
}

df = pd.DataFrame(datos)
df.to_csv("resultados_sentimientos.csv", index=False)

print("Archivo guardado correctamente.")
```


## 🎯 Conclusiones

- El modelo permite clasificar comentarios en positivos, negativos y neutrales.
- Puede utilizarse para analizar feedback de clientes de forma automática.
- Tiene aplicaciones directas en Customer Success, atención al cliente y análisis de experiencia del cliente.
